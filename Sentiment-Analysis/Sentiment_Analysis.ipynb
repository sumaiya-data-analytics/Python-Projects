```python
# step 1:importing liraries (these are default libraries which is downloaded along with python)
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#step 2:to understand the meaning of the sentence we call the library --textblob
from textblob import TextBlob

#step 3: to make the word cloud we need to call the libaray
from wordcloud import WordCloud
```

```python
# step 4:load the dataset
dataset=pd.read_csv("Sentiment_Analysis .csv")
# head-- it display the by default top 5 records from the file if we dont specify the number
dataset.head()
#dataset.head(8) - it show the top 8 from 0 to 7
```

```python
# Define the sentiment analysis functions def function_name(input) then next logic
def get_sentiment(sentence):
    meaning = TextBlob(sentence)

    # tone in python is called polarity-- itt check whether the sentence is positive or negative
    if meaning.sentiment.polarity >0:
        return "Positive"

    elif meaning.sentiment.polarity<0:
        return "Negative"

    else:
        return "Neutral"   
```

```python
# in sheet i want an column called sentiment and after each sentence it should show whether that particular sentence is pos,neg,neutral
dataset["Sentiment"]=dataset["Sentences"].apply(get_sentiment)
#filestoredname[new column to add name]=filestoredsheetname[column name already presentname]
```

```python
# to save the file
dataset.to_excel("Sentiment_analysis_sumaiya.xlsx",index=True)
```

```python
# i need to separte negative sentiment
negative_reviews=dataset[dataset["Sentiment"]=="Negative"]
```

```python
negative_reviews.head(10)
```

```python
# i need to make the paragraph of all the negative sentiment sentence
# so after that for that pargraph i can make wordcloud
negative_para = " ".join(negative_reviews["Sentences"])
print(negative_para)
```

```python
# create the wordcloud for the negative para (here we need to specify the dimension )
wordcloud = WordCloud(width=800,height=400,background_color="white").generate(negative_para)

# plot the wordcloud

plt.figure(figsize=(10,6))
plt.imshow(wordcloud,interpolation='bilinear')
plt.title("Most Common Words in Negative Sentences")
plt.axis("off")
plt.show()
```
